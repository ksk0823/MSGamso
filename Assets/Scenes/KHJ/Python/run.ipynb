# Unity 데이터 실시간 시각화 서버

이 노트북은 Unity에서 전송하는 데이터를 실시간으로 시각화하기 위한 TCP 서버를 구현합니다.

## 1. 필요한 라이브러리 및 모듈 임포트

```python
# 필요한 라이브러리 설치 (처음 한 번만 실행)
# %pip install matplotlib numpy
```

```python
# 구현된 모듈 임포트
from UnityGrapherApp import UnityGrapherApp

print("모듈 임포트 완료")
```

## 2. 애플리케이션 객체 생성 및 실행

```python
# 애플리케이션 객체 생성
app = UnityGrapherApp(host='127.0.0.1', port=5555)
print("Unity 데이터 시각화 애플리케이션 설정 완료")
```

## 3. 서버 실행

```python
# 서버 시작 (시각화는 하지 않고 백그라운드에서 실행)
print("Unity 데이터 시각화 서버 시작 중...")
app.server.start()
```

## 4. 시각화 시작

시각화 창을 시작합니다. 아래 셀을 실행하면 여러 개의 시각화 창을 띄울 수 있는 컨트롤 창이 나타납니다.

```python
# 시각화 시작
print("데이터 시각화 시작 중...")
app.visualizer.start()
```

## 5. 데이터 확인

현재 수집된 데이터를 확인합니다. 이 셀은 데이터를 확인하고 싶을 때마다 실행할 수 있습니다.

```python
# 마지막으로 수신한 데이터 출력
print("마지막으로 수신된 데이터:")
data = app.data_manager.get_data()
if data['last_data']:
    import json
    print(json.dumps(data['last_data'], indent=2))
else:
    print("수신된 데이터 없음")
```

## 6. 서버 종료하기

아래 셀을 실행하면 서버를 종료합니다. (필요할 때만 실행)

```python
# 서버 종료
app.stop()
print("서버를 종료했습니다.")
```
